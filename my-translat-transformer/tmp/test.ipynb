```python
import numpy as np
import matplotlib.pyplot as plt

def rasterize_polygon(vertices, resolution):
    # Find bounding box
    min_x = min(vertices, key=lambda vertex: vertex[0])[0]
    max_x = max(vertices, key=lambda vertex: vertex[0])[0]
    min_y = min(vertices, key=lambda vertex: vertex[1])[1]
    max_y = max(vertices, key=lambda vertex: vertex[1])[1]

    # Create grid
    grid = np.zeros((resolution, resolution))

    # Scanline Fill Algorithm
    for y in range(min_y, max_y + 1):
        intersections = []
        for i in range(len(vertices)):
            current = vertices[i]
            next = vertices[(i + 1) % len(vertices)]
            if current[1] <= y < next[1] or next[1] <= y < current[1]:
                x_intersection = int(current[0] + (float(y - current[1]) / (next[1] - current[1])) * (next[0] - current[0]))
                intersections.append(x_intersection)

        intersections.sort()
        for i in range(0, len(intersections), 2):
            for x in range(intersections[i], intersections[i + 1] + 1):
                grid[y][x] = 1

    return grid

def mark_partially_filled_cells(grid):
    for y in range(grid.shape[0]):
        row = grid[y]
        first_filled = -1
        for x in range(grid.shape[1]):
            if row[x] == 1:
                if first_filled == -1:
                    first_filled = x
            elif first_filled != -1:
                for px in range(first_filled, x):
                    grid[y][px] = 0.5
                first_filled = -1
        if first_filled != -1:
            for px in range(first_filled, grid.shape[1]):
                grid[y][px] = 0.5

    return grid

# Example usage
vertices = [(2, 2), (8, 2), (8, 8), (5, 10), (2, 8)]
resolution = 100

grid = rasterize_polygon(vertices, resolution)
grid = mark_partially_filled_cells(grid)
v = np.array(vertices)
plt.scatter(v[:,0], v[:,1])

# Display grid
plt.imshow(grid, cmap='gray', origin='lower')
plt.show()
```

```python
import math
import random

def generate_random_points_on_circle(n, radius, center=(0, 0)):
    points = []
    for _ in range(n):
        angle = random.uniform(0, 2*math.pi)
        x = center[0] + radius * math.cos(angle)
        y = center[1] + radius * math.sin(angle)
        points.append((x, y))
    return points

# Example usage
num_points = 5
circle_radius = 10
circle_center = (0, 0)

random_points = generate_random_points_on_circle(num_points, circle_radius, circle_center)
print(random_points)
```

```python
def load_velocity(flow_dir):
    all_u_mat = np.load(flow_dir +'all_u_mat.npy')
    all_ui_mat = np.load(flow_dir +'all_ui_mat.npy')
    all_v_mat = np.load(flow_dir +'all_v_mat.npy' )
    all_vi_mat = np.load(flow_dir +'all_vi_mat.npy')
    all_Yi = np.load(flow_dir +'all_Yi.npy' )
    vel_field_data = [all_u_mat, all_v_mat, all_ui_mat, all_vi_mat, all_Yi]
    return vel_field_data
```

```python
def extract_velocity(vel_field_data, t, rzn):
    nmodes =  vel_field_data[2].shape[1]
    vx = vel_field_data[0][t,:,:]
    vy = vel_field_data[1][t,:,:] 

    for m in range(nmodes):
        vx += vel_field_data[2][t, m, :, :]*vel_field_data[4][t, rzn,m]
        vy += vel_field_data[3][t, m, :, :] * vel_field_data[4][t, rzn,m]

    return np.stack([vx,vy], axis=0)
```

```python
def load_vel(data_path):
    all_u_mat = np.load(data_path +'all_u_mat.npy')
    all_ui_mat = (np.load(data_path +'all_ui_mat.npy'))
    all_v_mat = (np.load(data_path +'all_v_mat.npy'))
    all_vi_mat = (np.load(data_path +'all_vi_mat.npy'))
    all_Yi = np.load(data_path +'all_Yi.npy')
    obstacle_mask = (np.load(data_path + 'obstacle_mask.npy'))
    vel_field_data = [all_u_mat, all_v_mat, all_ui_mat, all_vi_mat, all_Yi, obstacle_mask]
    return vel_field_data
```

```python
def extract_velocity(vel_field_data, t, rzn, as_image=False):
    nmodes =  vel_field_data[2].shape[1]
    vx = vel_field_data[0][t,:,:].copy()
    vy = vel_field_data[1][t,:,:].copy()
    obstacle_mask = vel_field_data[5][t,:,:] 
    obstacle_mask = np.float32(obstacle_mask)

    for m in range(nmodes):
        vx += vel_field_data[2][t, m, :, :]*vel_field_data[4][t, rzn,m]
        vy += vel_field_data[3][t, m, :, :] * vel_field_data[4][t, rzn,m]

    if as_image:
        im = np.stack([vx, vy, obstacle_mask], axis=0)
        return im
    else:
        return vx, vy
```

```python
def plot_vel_field(U, V, Ui,Vi, Yi,ax, t,r=0, g_strmplot_lw=1, g_strmplot_arrowsize=1):
    # Make modes the last axis
    Ui = np.transpose(Ui,(0,2,3,1))
    Vi = np.transpose(Vi,(0,2,3,1))
    vx_grid = U[t,:,:] + np.dot(Ui[t,:,:,:],Yi[t,r,:])
    vy_grid = V[t,:,:] + np.dot(Vi[t,:,:,:],Yi[t,r,:])
    vx_grid = np.flipud(vx_grid)
    vy_grid = np.flipud(vy_grid)
    Xs = np.arange(0,100) 
    Ys = np.arange(0,100) 
    X,Y = np.meshgrid(Xs, Ys)
    ax.streamplot(X, Y, vx_grid, vy_grid, color = 'grey', zorder = 0,  linewidth=g_strmplot_lw, arrowsize=g_strmplot_arrowsize, arrowstyle='->')
    v_mag_grid = (vx_grid**2 + vy_grid**2)**0.5
    im = ax.contourf(X, Y, v_mag_grid, cmap = "Blues", alpha = 0.5, zorder = -1e5)
    return im
```

```python
def setup_ax(ax, show_xlabel= True, 
                        show_ylabel=True, 
                        show_xticks=True,
                        show_yticks=True,
                        lab_fs = 15,
                        tick_fs = 14,
                        ):
    ax.set_aspect('equal', adjustable='box')
    ax.set_xlim([0,100])
    ax.set_ylim([0,100])
    xticks = np.arange(0,100 +25,25)
    yticks = xticks.copy()
    ax.xaxis.set_ticks(xticks)
    ax.yaxis.set_ticks(yticks)
    ax.set_xticklabels(xticks, fontsize=tick_fs)
    ax.set_yticklabels(yticks, fontsize=tick_fs)

    if not show_xticks:
        ax.tick_params(axis='x',       
                        which='both',      
                        bottom=False,      
                        labelbottom=False,
                        labelsize=tick_fs)
    if not show_yticks:
        ax.tick_params(axis='y',       
                which='both',      
                left=False,      
                labelleft=False,
                labelsize=tick_fs)
    xlabel = f"X "
    ylabel = f"Y "
    if show_xlabel:
        ax.set_xlabel(xlabel, fontsize=lab_fs)
    if show_ylabel:
        ax.set_ylabel(ylabel,fontsize=lab_fs)
```

```python
def plot_velocity_field(vx, vy, path):
    X, Y = np.meshgrid(np.linspace(0, 50, 50), np.linspace(0, 50, 50))
    # vx = np.nan_to_num(vx, nan=0.0)
    # vy = np.nan_to_num(vy, nan=0.0)
    mag = np.sqrt(vx**2 + vy**2)
    vmin = np.min(mag)
    vmax = np.max(mag)
    print(np.max(mag))
    # plt.cla()
    plt.streamplot(X, Y, vx, vy)
    plt.contourf(X, Y, mag, cmap='Blues')
    plt.clim(vmin, vmax)
    plt.colorbar()
    plt.title('t=0')
    plt.savefig(path+"/bob_0", dpi=300)
```

```python
data_path = "/media/HDD/rohit/Translation_transformer/my-translat-transformer/data/BoB/processed_datasets/50x50"
desired_mag = 0.4
```

```python

def select_env(all_vxvy_interpolated, y_start, y_end, x_start, x_end, run_name):
    rzn = all_vxvy_interpolated[:, :, y_start:y_end, x_start:x_end]
    vel_mag = np.sqrt(rzn[0]**2 + rzn[1]**2)
    print('Old_mag', np.max(vel_mag))
    max_vel_mag_t = np.max(vel_mag, axis=(1,2)).reshape(-1, 1, 1)
    vx = rzn[0]* (desired_mag / max_vel_mag_t)
    vy = rzn[1]* (desired_mag / max_vel_mag_t)
    vel_mag = np.sqrt(vx**2 + vy**2)
    print('New_mag',np.max(vel_mag))
    plot_velocity_field(vx[0], vy[0], data_path+run_name)
    rzn = np.stack([vx, vy], axis=0)
    path = data_path+run_name+'/all_vxvy.npy'
    print('Vel_shape', rzn.shape)
    np.save(path, rzn)
```

```python
def are_arrays_unique(arrays):
    # Convert each array to a tuple for immutability
    array_tuples = [tuple(arr.flatten()) for arr in arrays]
    
    # Check if all tuples are unique
    if len(set(array_tuples)) != len(arrays):
        return False
    
    # Optionally, check for elementwise equality
    for i in range(len(arrays)):
        for j in range(i + 1, len(arrays)):
            if np.array_equal(arrays[i], arrays[j]):
                return False
    
    return True
if are_arrays_unique(a):
    print("All arrays are unique and not equal elementwise.")
else:
    print("There are duplicate arrays or arrays that are equal elementwise.")
```
